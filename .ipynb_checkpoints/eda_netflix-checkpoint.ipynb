# Exploratory Data Analysis on Netflix Movies and TV Shows

**Netflix** is one of the largest streaming platforms, offering a wide variety of movies and TV shows across different countries and genres.  
This project focuses on performing exploratory data analysis (**EDA**) on Netflix's content dataset to understand content distribution, trends over time, and patterns across genres, countries, and release years.

The analysis includes data cleaning, preprocessing, visualization, and answering analytical questions using Python.
