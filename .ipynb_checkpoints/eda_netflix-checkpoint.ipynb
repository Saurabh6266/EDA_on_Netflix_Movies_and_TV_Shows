## Data Cleaning and Feature Engineering

**Let's start by checking missing values because, in general, the industrial datasets are too messy and unstructured, so we should clean the data before visualizing anything.**

```python
df.isnull().sum().sort_values(ascending=False)
```

The dataset contains missing values, especially 'director' contains **2634** followed by 'country' **831**, 'cast' **825** and so on. It happens when the provider does not provide the actual and real information for some sort of reason and **Pandas** directly convert those missing or unknown values into **NaN** which stands for **Not a number**.

**Let's visualize the missing values with a bar plot.**

```python
missing = df.isnull().sum().sort_values(ascending=False)

plt.figure(figsize=(12,6))
sns.barplot(x=missing.values, y=missing.index)
plt.title("Missing Values by Column")
plt.xlabel("Number of Missing Values")
plt.ylabel("Columns")
```

**Check whether duplicate records exist in the dataset.**

```python
df.duplicated().sum()
```

**Handle missing values**

```python
df["director"] = df["director"].fillna("Unknown")
df["cast"] = df["cast"].fillna("Unknown")
df["country"] = df["country"].fillna("Unknown")
```

**The `date_added` column is converted to a datetime format to enable time-based analysis, such as extracting the year and month when content was added to Netflix.** 

```python
df["date_added"] = pd.to_datetime(df["date_added"], errors="coerce")

df["year_added"] = df["date_added"].dt.year
df["month_added"] = df["date_added"].dt.month
```

```python
df.head()
```

**What will be the content age?**

```python
df["content_age"] = df["year_added"] - df["release_year"]
```

Content age represents the number of years between the original release and the date the content was added to Netflix. This helps analyze whether Netflix prefers recently released titles or older catalog content.

## Exploratory Data Analysis

### 1. What is the distribution of Movies vs TV Shows? 

```python
df.type.value_counts()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x="type", data=df, palette="Set2")
plt.title("Distribution of Movies and TV Shows on Netflix")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")
```

Movies constitute the majority of titles available on Netflix.

### 2. How has Netflix content grown over time?

```python
year_counts = df["year_added"].value_counts().sort_index()

plt.figure(figsize=(12,6))
plt.plot(year_counts.index, year_counts.values, marker='o')
plt.title("Growth of Netflix Content Over Time")
plt.xlabel("Year Added")
plt.ylabel("Number of Titles Added")

plt.grid(True)
```

It's like netflix's content didn't had a growth till 2014 and after that it suddenly rose with rapid growth but in around 2019 it decreased again.

### 3. Which countries produce the most content?

```python
top_countries = df["country"].value_counts().head(10)
top_countries
```

```python
sns.barplot(x=top_countries.values, y=top_countries.index);
```

It seems like the United States dominated with a large amount of content produced and then followed by India, United Kingdom, etc.

### 4. What are the most common ratings?

```python
df["rating"].value_counts()
```

```python
sns.countplot(y="rating", data=df);
```

### 5. What genres dominate Netflix?

```python
genres = df.assign(genre=df['listed_in'].str.split(',')).explode('genre')
```

```python
top_genres = genres['genre'].value_counts().head(10)

plt.figure(figsize=(10,6))

sns.barplot(
    x=top_genres.values,
    y=top_genres.index,
    palette="viridis"
)

plt.title("Top 10 Genres on Netflix")
```

### 6. Does Netflix favor newer or older content?

```python
plt.figure(figsize=(10,6))

sns.histplot(df["content_age"], bins=30, kde=True);

plt.title("Distribution of Content Age")
plt.xlabel("Years Between Release and Netflix Addition")
plt.ylabel("Frequency")
```

### 7. Let's compare line plot and bar plot for content growth

```python
sns.lineplot(x=year_counts.index, y=year_counts.values);
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    x=year_counts.index.astype(int),
    y=year_counts.values,
    color="skyblue"
);

plt.xticks(rotation=45)
```

For time trends, line plots are more appropriate than bar plots.
